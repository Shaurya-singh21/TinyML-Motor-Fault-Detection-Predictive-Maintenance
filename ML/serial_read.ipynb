```python
import serial
import numpy as np
import pandas as pd

ser = serial.Serial("/dev/ttyACM0", 115200, timeout=2)
ser.reset_input_buffer()

data = []

for i in range(500):
    raw = ser.read(68)

    if len(raw) == 68:
        
        features = np.frombuffer(raw[:60], dtype="<f4")
        prediction = np.frombuffer(raw[60:64], dtype="<i4")[0]
        majority = np.frombuffer(raw[64:68], dtype="<i4")[0]
        print(features)

        data.append([*features,prediction,majority])

ser.close()

data = np.vstack(data)
columns = [
    "Variance_X", "Variance_Y", "Variance_Z",
    "RMS_X", "RMS_Y", "RMS_Z",
    "PtP_X", "PtP_Y", "PtP_Z",
    "Dominant_X", "Dominant_Y", "Dominant_Z",
    "Spec_Crest_X", "Spec_Crest_Y", "Spec_Crest_Z",
    "Prediction",
    "Majority_Vote"
]
df = pd.DataFrame(data, columns=columns)
df.to_csv("healthy_500_windows.csv", index=False)

print(df.shape)
```
